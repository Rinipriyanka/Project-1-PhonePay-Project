```python
#Aggregate Transaction

path="/content/pulse/data/aggregated/transaction/country/india/state/"
Agg_Trans_state_list=os.listdir(path)
Agg_Trans_state_list
```

```python
Agg_Trans_Dict={
    "State":[],"Year":[],"Quarter":[],"Transaction_Type":[],"Transaction_Count":[]
}

for i in Agg_Trans_state_list:
  p_i=path+i+"/"
  Agg_yr=os.listdir(p_i)
  for j in Agg_yr:
    p_j=p_i+j+"/"
    Agg_yr_list=os.listdir(p_j)
    for k in Agg_yr_list:
      p_k=p_j+k
      with open(p_k,"r")as file:
        D=json.load(file)
        for z in D['data']['transactionData']:
            Name = z['name']
            Count = z['paymentInstruments'][0]['count']
            Agg_Trans_Dict['State'].append(i)
            Agg_Trans_Dict['Year'].append(int(j))
            Agg_Trans_Dict['Quarter'].append(int(k.strip('.json')))
            Agg_Trans_Dict['Transaction_Type'].append(Name)
            Agg_Trans_Dict['Transaction_Count'].append(Count)
Agg_trans=pd.DataFrame(Agg_Trans_Dict)
Agg_trans.head(10)
```

```python
#Aggregate User Data

path="/content/pulse/data/aggregated/user/country/india/state/"
Agg_user_list=os.listdir(path)
Agg_user_list
```

```python
Agg_user_dict={
    "State":[],"Year":[],"Quarter":[],"Registered_User":[]
}

for i in Agg_user_list:
  p_i=path+i+"/"
  Agg_state=os.listdir(p_i)
  for j in Agg_state:
    p_j=p_i+j+"/"
    Agg_year=os.listdir(p_j)
    for k in Agg_year:
      p_k=p_j+k
      with open(p_k,"r") as file:
        D=json.load(file)
        Registered_users=D["data"]["aggregated"]["registeredCount"]
        Agg_user_dict["State"].append(i)
        Agg_user_dict["Year"].append(int(j))
        Agg_user_dict["Quarter"].append(k.replace(".json",""))
        Agg_user_dict["Registered_User"].append(Registered_users)
Agg_user=pd.DataFrame(Agg_user_dict)
Agg_user.head(10)
```

```python
#Map Transaction

path="/content/pulse/data/map/transaction/hover/country/india/state/"
Map_trans_list=os.listdir(path)
Map_trans_list
```

```python
Map_trans_Dict={
    "State":[],"Year":[],"Quarter":[],"District_Name":[],"Transaction_count":[],"Transaction_amount":[]}

for i in Map_trans_list:
  p_i=path+i+"/"
  Map_state=os.listdir(p_i)
  for j in Map_state:
    p_j=p_i+j+"/"
    Map_yr=os.listdir(p_j)
    for k in Map_yr:
      p_k=p_j+k
      with open(p_k,"r") as file:
        D=json.load(file)
        for z in D["data"]["hoverDataList"]:
            District = z['name']
            Count = z['metric'][0]['count']
            Amount = z['metric'][0]['amount']
            Map_trans_Dict['State'].append(i)
            Map_trans_Dict['Year'].append(int(j))
            Map_trans_Dict['Quarter'].append(int(k.replace('.json', '')))
            Map_trans_Dict['District_Name'].append(District)
            Map_trans_Dict['Transaction_count'].append(Count)
            Map_trans_Dict['Transaction_amount'].append(Amount)
Map_trans=pd.DataFrame(Map_trans_Dict)
Map_trans

```

```python
# Top User

path="/content/pulse/data/top/user/country/india/state/"
Top_user_list=os.listdir(path)
Top_user_list
```

```python
Top_user_Dict={"State":[],"Year":[],"Quarter":[],"District_Name":[],"Registered_Users":[]}

for i in Top_user_list:
    p_i = path + i + "/"
    Top_yr = os.listdir(p_i)

    for j in Top_yr:
        p_j = p_i + j + "/"
        Top_yr_list = os.listdir(p_j)

        for k in Top_yr_list:
            p_k = p_j + k

            with open(p_k, "r") as file:
                D = json.load(file)

                for z in D['data']['districts']:

                    District = z['name']
                    Registered_Users = z['registeredCount']

                    Top_user_Dict['State'].append(i)
                    Top_user_Dict['Year'].append(int(j))
                    Top_user_Dict['Quarter'].append(int(k.replace('.json', '')))
                    Top_user_Dict['District_Name'].append(District)
                    Top_user_Dict['Registered_Users'].append(Registered_Users)
Top_user=pd.DataFrame(Top_user_Dict)
Top_user.head(10)
```
